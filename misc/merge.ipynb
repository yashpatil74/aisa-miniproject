```python
import pandas as pd

df_res = pd.read_csv("BVBRC_genome_resistant.csv")
df_sus = pd.read_csv("BVBRC_genome_susceptible.csv")

df_combined = pd.concat([df_res, df_sus], ignore_index=True)

df_combined.to_csv("metadata_combined.csv", index=False)

print(df_combined['Resistant Phenotype'].value_counts())
```

```python

df = pd.read_csv("metadata_combined.csv")

# Drop any accidental duplicates from the merge process
df_clean = df.drop_duplicates(subset=["Genome ID"])

# Extract the IDs, convert to string, and save to a clean text file
df_clean["Genome ID"].astype(str).to_csv("genome_ids.txt", index=False, header=False)

print(f"Extraction complete! Ready to download {len(df_clean)} genomes.")
```
